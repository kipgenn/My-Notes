Main structure:
Head -> Body -> Footer

Head should contain only needed by the maching like title(tab name)

Body: header -> main 

block elements: div, h, 
inline elements: span

CSS:
* {
    margin: 0;
    padding: 0;
    box-sizing: border-box; 
}


flex on parent makes all children flexitems kind of like inline
horizontal axis: justify-content, values: flex-start(all items in left), center(all in centre), space-between(first and last at extreme others spaced evenly), space-evenly(extreme touch boundary with half the size of spaced evenly in between), space-around (all items spaced same)

vertical: align items, flex-start, flex-end , center, stretch(force fill all height)
